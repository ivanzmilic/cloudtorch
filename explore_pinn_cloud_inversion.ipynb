```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
file = '/dat/milic/MiHI_filament/inv_results/fits_and_params.npz'
```

```python
output = np.load(file)
```

```python
spectra_fits = output['fits']
spectra_obs = output['obs']
params = output['params']
```

```python
spectra_fits.shape
spectra_obs.shape
params.shape
```

```python
# Parameter legend:
 # [S1, dtau1, vlos1, dv1, loga1, S2, dtau2, vlos2, dv2, loga2].
 # S: source function
 # dtau: optical depth
 # vlos: line-of-sight velocity
 # dv: Doppler width
 # loga: logarithm of the damping parameter - this one is if I remember correctly not fit.
```

```python
# Plot some random spectra and their fits
num_spectra = spectra_fits.shape[0]*spectra_fits.shape[1]*spectra_fits.shape[2]
random_indices = np.random.choice(num_spectra, size=5, replace=False)

# Use subplots to plot these 5 test fits: 
plt.figure(figsize=(15, 3))
for i, idx in enumerate(random_indices):
    plt.subplot(1, 5, i+1)
    plt.plot(spectra_obs.reshape(-1, spectra_obs.shape[-1])[idx], label='Observed')
    plt.plot(spectra_fits.reshape(-1, spectra_fits.shape[-1])[idx], label='Fitted')
```

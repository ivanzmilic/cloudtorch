```python
import numpy as np 
import matplotlib.pyplot as plt 
from astropy.io import fits 
```

```python
stokes = fits.open("~/data/img_fit.210800..211599.00.00.lr0250.cube.fits")[0].data
```

```python
ll = fits.open("~/data/img_fit.210800..211599.00.00.lr0250.cube.fits")[1].data
```

```python
stokes.shape
```

```python
plt.imshow(np.mean(stokes[10:-10,10:-10,0,95:105],axis=2), origin='lower', cmap='gray')
plt.title("Far blue wing of Stokes I")
plt.colorbar()
plt.savefig('mihi_I.png', bbox_inches='tight', dpi=300)
```

```python
plt.figure(figsize=(8,6))
plt.plot(ll[:-70],stokes[30,90,0,:-70], label = 'x=90, y=30')
plt.plot(ll[:-70],stokes[80,80,0,:-70], label = 'x=80, y=80')
plt.plot(ll[:-70],stokes[20,40,0,:-70], label = 'x=40, y=20')
plt.legend()
plt.savefig('mihi_I_spectra.png', bbox_inches='tight', dpi=300)
```
